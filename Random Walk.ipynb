```python
import numpy as np
import matplotlib.pyplot as plt
import random
import math

def randomwalk1D(n):
    y = 0
    positions = [y]
    directions = ["UP","DOWN"]
    for i in range(1,n+1):
        step = random.choice(directions)
        if step == "UP":
            y += 1
        elif step == "DOWN":
            y -= 1
        positions.append(y) #mutation
    #so here ith index of position is the position at ith time. 
    return positions

#so here let's generalize the notion of a random walk to d-directions where d is a positive integer.  
#we won't use a lot of storage. It is a just a sequence
def randomwalk(n,d):
    sequence = [0]
    for i in range(1,n+1):
        step = random.randint(1,d)
        sequence.append(step)
        #so the ith index element in the sequence represents the decision taken at t = i.
    return (sequence,d)

#from the output, we cannot tell the dimension etc. Might be important. What about visualising these walks? 
#we don't need tiling by polygons?

def randomwalkembedding(n,d):
    sequence = randomwalk(n,d)[0]
    position = (0,0)
    walk = [(0,0)]
    vector = [(1,0)]
    for i in range(2,d+1):
        vector.append((math.cos((i-1)*2*(math.pi)/d), math.sin((i-1)*2*(math.pi)/d)))
    for i in range(1,n+1):
        position = tuple(x+y for x,y in zip(position, vector[sequence[i]-1]))
        walk.append(position)
    return walk

walk = randomwalkembedding(1000,10000)
x,y = zip(*walk)
plt.plot(x,y, linestyle = '-', color = 'green')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()

#This looks like some sort of brownian motion.
```
